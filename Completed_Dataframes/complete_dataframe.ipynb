```python
import pandas as pd

spread_2018 = pd.read_csv("../2018_spreads/2018_playoffs_spreads.csv")
spread_2018['Year'] = '2018'
spread_2019 = pd.read_csv("../2019_spreads/2019_playoffs_spreads.csv")
spread_2019['Year'] = '2019'
spread_2020 = pd.read_csv("../2020_spreads/2020_playoffs_spreads.csv")
spread_2020['Year'] = '2020'
spread_2021 = pd.read_csv("../2021_spreads/2021_playoffs_spreads.csv")
spread_2021['Year'] = '2021'
spread_2022 = pd.read_csv("../2022_spreads/2022_playoffs_spreads.csv")
spread_2022['Year'] = '2022'
```

```python
# Combine them
spread_df = pd.concat(
    [spread_2018, spread_2019, spread_2020, spread_2021, spread_2022],
    ignore_index=True
)
```

```python
spread_df
```

```python
# Ensure Date is 4 digits and Year is int
spread_df['Date'] = spread_df['Date'].astype(str).str.zfill(4)
spread_df['Year'] = spread_df['Year'].astype(str)
```

```python
# Combine into full date string
spread_df['GameDate'] = pd.to_datetime(spread_df['Year'] + spread_df['Date'], format='%Y%m%d', errors='coerce')
```

```python
spread_df
```

```python
team_name_to_abbreviation = {
    "Atlanta": "ATL",
    "Boston": "BOS",
    "Brooklyn": "BKN",
    "Charlotte": "CHA",
    "Chicago": "CHI",
    "Cleveland": "CLE",
    "Dallas": "DAL",
    "Denver": "DEN",
    "Detroit": "DET",
    "GoldenState": "GSW",
    "Houston": "HOU",
    "Indiana": "IND",
    "LAClippers": "LAC",
    "LALakers": "LAL",
    "Memphis": "MEM",
    "Miami": "MIA",
    "Milwaukee": "MIL",
    "Minnesota": "MIN",
    "NewOrleans": "NOP",
    "NewYork": "NYK",
    "OklahomaCity": "OKC",
    "Orlando": "ORL",
    "Philadelphia": "PHI",
    "Phoenix": "PHX",
    "Portland": "POR",
    "Sacramento": "SAC",
    "SanAntonio": "SAS",
    "Toronto": "TOR",
    "Utah": "UTA",
    "Washington": "WAS"
}
```

```python
spread_df['TEAM_ABBREVIATION'] = spread_df['Team'].str.replace(" ", "").map(team_name_to_abbreviation)
spread_df
```

```python
spread_df['GameDate'].info
```

```python
spread_df['HOME_AWAY'] = spread_df['VH'].map({'H': 'Home', 'V': 'Away'})
spread_df
```

```python
team_games_merged = pd.read_csv('../Merges/team_with_game_metadata.csv')
team_games_merged
```

```python
team_games_merged['GAME_DATE'] = pd.to_datetime(team_games_merged['GAME_DATE'])
```

```python
team_with_spreads = pd.merge(
    team_games_merged,
    spread_df,
    how='inner',
    left_on=['TEAM_ABBREVIATION', 'GAME_DATE', 'HOME_AWAY'],
    right_on=['TEAM_ABBREVIATION', 'GameDate', 'HOME_AWAY'],
    
)
```

```python
team_with_spreads
```

```python
team_with_spreads.columns
```

```python
team_with_spreads.drop(columns=['GameDate'], inplace=True)
team_with_spreads.drop(columns=['Date'], inplace=True)
team_with_spreads
```

```python
# Step 1: Standardize matchup strings so both teams have the same series ID
def get_series_id(row):
    teams = sorted(row['MATCHUP'].replace('@', 'vs.').replace('vs.', '').split())
    return f"{teams[0]}_vs_{teams[1]}_{row['SEASON_ID']}"

team_with_spreads['SERIES_ID'] = team_with_spreads.apply(get_series_id, axis=1)

# Step 2: Sort by date to ensure cumulative logic works
team_with_spreads = team_with_spreads.sort_values(by=['SERIES_ID', 'GAME_DATE'])

# Step 3: Create SERIES_WINS: cumulative wins per team in each series
team_with_spreads['SERIES_WINS'] = team_with_spreads.groupby(['SERIES_ID', 'TEAM_ABBREVIATION'])['WL'].transform(lambda x: (x == 'W').cumsum())
team_with_spreads
```

```python
# Step 1: Create a key to represent each unique game instance (date + series)
team_with_spreads['GAME_KEY'] = (
    team_with_spreads['SERIES_ID'] + '_' + team_with_spreads['GAME_DATE'].astype(str)
)

# Step 2: Select relevant columns to self-merge for opponent lookup
opponent_wins_df = team_with_spreads[['GAME_KEY', 'TEAM_ABBREVIATION', 'SERIES_WINS']]
opponent_wins_df = opponent_wins_df.rename(columns={
    'TEAM_ABBREVIATION': 'OPPONENT_ABBREVIATION',
    'SERIES_WINS': 'OPPONENT_SERIES_WINS'
})

# Step 3: Merge to get opponent's SERIES_WINS
team_with_spreads = pd.merge(
    team_with_spreads,
    opponent_wins_df,
    on='GAME_KEY',
    how='left'
)

# Step 4: Remove rows where team and opponent are the same
team_with_spreads = team_with_spreads[
    team_with_spreads['TEAM_ABBREVIATION'] != team_with_spreads['OPPONENT_ABBREVIATION']
]

# Step 5: Create SERIES_STANDING string like "2–1"
team_with_spreads['SERIES_STANDING'] = (
    team_with_spreads['SERIES_WINS'].astype(str) + '–' +
    team_with_spreads['OPPONENT_SERIES_WINS'].astype(str)
)
team_with_spreads
```

```python
team_with_spreads.to_csv('team_with_spreads.csv', index=False)
```
