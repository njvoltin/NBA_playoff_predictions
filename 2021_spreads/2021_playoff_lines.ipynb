```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
# Set URL and headers
url = "https://www.sportsbookreviewsonline.com/scoresoddsarchives/nba-odds-2020-21/"
headers = {
    "User-Agent": "Mozilla/5.0"
}
```

```python
# Get and parse the page
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')
```

```python
# Find the table by class name
table = soup.find("table", class_="table bg-white table-hover table-bordered table-sm")
```

```python
# Read the HTML table into a DataFrame
df = pd.read_html(str(table))[0]
df.columns = df.iloc[0]
df = df.drop(index=0).reset_index(drop=True)
```

```python
# Preview and clean (optional)
df.columns = df.columns.droplevel(0) if isinstance(df.columns, pd.MultiIndex) else df.columns
df = df.dropna(how='all')  # Drop rows that are completely empty
```

```python
# Display
df.head()
```

```python
# First, make sure the Date column is treated as an integer
df["Date"] = df["Date"].astype(int)

# Filter for playoff dates
df = df[(df["Date"] >= 522) & (df["Date"] <= 722)]

# Preview filtered DataFrame
df.head()
```

```python
df.to_csv("2021_playoffs_spreads.csv", index=False)
```
