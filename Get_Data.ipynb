```python
import pandas as pd
import time
from tqdm.notebook import tqdm
from nba_api.stats.endpoints import LeagueGameLog
```

```python
# Set seasons range (2010 - 2024)
seasons = [f"{year-1}-{str(year)[-2:]}" for year in range(2010, 2025)]
```

```python
# Initialize lists to store DataFrames
all_team_logs = []
all_player_logs = []
```

```python
# Pull Team Game Logs
print("Starting Team Game Logs Collection...\n")
for season in tqdm(seasons, desc="Fetching TEAM Data"):
    try:
        team_log = LeagueGameLog(
            season=season,
            season_type_all_star='Playoffs',
            player_or_team_abbreviation='T'
        )
        team_df = team_log.get_data_frames()[0]
        team_df['SEASON'] = season
        all_team_logs.append(team_df)
        time.sleep(1.5)  # Respect NBA API rate limits
    except Exception as e:
        print(f"Error fetching TEAM data for {season}: {e}")
```

```python
# Combine and save Team Logs
nba_playoffs_team_games = pd.concat(all_team_logs, ignore_index=True)
nba_playoffs_team_games.to_csv('nba_playoffs_team_games_2010_2024.csv', index=False)
print("\n✅ Team game logs saved as 'nba_playoffs_team_games_2010_2024.csv'\n")
```

```python
# Pull Player Game Logs
print("Starting Player Game Logs Collection...\n")
for season in tqdm(seasons, desc="Fetching PLAYER Data"):
    try:
        player_log = LeagueGameLog(
            season=season,
            season_type_all_star='Playoffs',
            player_or_team_abbreviation='P'
        )
        player_df = player_log.get_data_frames()[0]
        player_df['SEASON'] = season
        all_player_logs.append(player_df)
        time.sleep(1.5)  # Respect NBA API rate limits
    except Exception as e:
        print(f"Error fetching PLAYER data for {season}: {e}")
```

```python
# Combine and save Player Logs
nba_playoffs_player_stats = pd.concat(all_player_logs, ignore_index=True)
nba_playoffs_player_stats.to_csv('nba_playoffs_player_stats_2010_2024.csv', index=False)
print("\n✅ Player game logs saved as 'nba_playoffs_player_stats_2010_2024.csv'\n")
```

```python
# Create Game Metadata File
print("Creating Game Metadata file...\n")
nba_playoffs_games = nba_playoffs_team_games[['GAME_ID', 'GAME_DATE', 'MATCHUP', 'TEAM_ID', 'TEAM_ABBREVIATION', 'WL', 'SEASON']]
nba_playoffs_games = nba_playoffs_games.drop_duplicates()
```

```python
nba_playoffs_games.to_csv('nba_playoffs_games_2010_2024.csv', index=False)
print("\n✅ Game metadata saved as 'nba_playoffs_games_2010_2024.csv'\n")
```
